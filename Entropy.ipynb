```python
# 2. Column-wise entropy function
# =========================

def column_entropy(alignment):
    n_cols = alignment.get_alignment_length()
    entropies = []

    for i in range(n_cols):
        column = alignment[:, i]
        counts = Counter(column)

        # remove gaps (recommended)
        if '-' in counts:
            del counts['-']

        # skip fully gapped columns
        if len(counts) == 0:
            entropies.append(np.nan)
            continue

        freqs = np.array(list(counts.values()), dtype=float)
        freqs = freqs / freqs.sum()

        entropies.append(entropy(freqs, base=2))

    return np.array(entropies)
```
